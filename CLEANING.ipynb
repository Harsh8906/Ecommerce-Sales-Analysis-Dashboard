```python
# Import required libraries
import pandas as pd 
import numpy as np 
```

```python
# Load Dataset
df = pd.read_excel('Ecommerce_Unclean_Project.xlsx')
```

```python
# Explore the dataset
# view first 5 records
df.head()
```

```python
# view dataset ingormation 
df.info()
```

```python
# statistical summary 
df.describe()
```

```python
# dataset shape 

df.shape
```

```python
# check missing values

df.isnull().sum()
```

```python
# check duplicate records

df.duplicated().sum()
```

```python
# remove duplicate records 

df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# replace invalid values 
# replace common invalid values with NaN 

df.replace(['N/A','NULL',''],np.nan,inplace=True)
```

```python
df
```

```python
# remove extra spaces

df = df.apply(lambda x : x.str.strip() if x.dtype == 'string'  else x)
```

```python
# standardize text fformat
# convert text into proper case.

df['Customer_Name'] = df['Customer_Name'].str.title()

df['City'] = df['City'].str.title()

df['State'] = df['State'].str.title()
```

```python
# validate email addresses
# keep only valid email records. 

EmailCheck = df['Email'].str.contains('@' , na = False)
```

```python
# convert data columns 

df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
df.info()
```

```python
# convert numeric columns

cols = ['Qty','Unit_Price','Discount']

for i in cols:
    df[i] = pd.to_numeric(df[i],errors='coerce')
```

```python
# remove invalid quantity 
# remove records having negative or zero quantity. 

df=df[df['Qty']>0]
```

```python
# fill missing values

df['Discount']=df['Discount'].fillna(0)
df['Phone']=df['Phone'].fillna('Unknown')
df['Delivery_Date']=df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# feature engineering 
# total sales 

df['Sales'] = df['Qty']*df['Unit_Price']
```

```python
df
```

```python
#net amount

df['Net_Amount'] = df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
# profit (example : 20%)

df['Profit'] = df['Net_Amount']*0.20
```

```python
df
```

```python
# Order month and year

df['Month']=df['Order_Date'].dt.month_name()
df['Year']=df['Order_Date'].dt.year
```

```python
df
```

```python
# weekday 

df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
df
```

```python
# final data validation

df.info()
```

```python
# export clean data 

df.to_csv('Clean_Ecommerce.csv',index=False)
```
